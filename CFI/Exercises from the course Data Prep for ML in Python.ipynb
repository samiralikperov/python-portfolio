# Cleaning Data & Imputation

### Exercise 1 - Removing Unwanted Strings

Once again, we are working with the phone dataset. However, as you know we identified a few problems.

Task:
- Explore the unique values in the marketplace column, identify the typos, and fix them.

```python
# Import data to a pandas dataframe
df_phone = pd.read_csv('phone_marketplace_dataset_cleaning_set.csv')
df_phone
```

```python
# Explore the unique values in the marketplace column
un_val = df_phone['marketplace'].value_counts()
un_val
```

```python
#Replace or remove all ! in the marketplace column
df_phone['marketplace'] = df_phone['marketplace'].str.replace('!','')
df_phone['marketplace'].value_counts()
```

### Exercise 2 - Imputing Nulls with Appropriate Values

Identify the column that has null values and populate the values using a basic imputation method of your choice.

Task:
- Identify the columns that need to be dealt with
- Populate missing values with a basic imputation method of your choice

```python
df_phone.isna().sum()
```

```python
# Check the rows that are null
df_null = df_phone[df_phone['price'].isnull()]
df_null
```

```python
#Find phone price averages (returned as series)
iphone_avg = df_phone.groupby('name') ['price'].mean()
iphone_avg
```

```python
df_phone['price'] = np.where(np.isnan(df_phone['price']), iphone_avg.loc[df_phone['name']], df_phone['price'])
df_phone
```

```python
# Checking to see if anymore null values are left
df_phone.isna().sum()
```

### Exercise 2 (Basic) - Visuals for Numeric Features

Create two visuals that displays the distribution of the starting price of cars based on transmission type. Then find the average starting price for each transmission type

Task:
- Plot the distribution of starting price for each transmission type in a histogram
- Print the average starting price for each transmission type

```python
# plot the distribution for manual transmission

histplot = sns.histplot(data = df_cars[df_cars['transmission_type']=='Manual'], x ='starting_price', bins = 10)
```

```python
# plot the distribution for automatic transmission

histplot = sns.histplot(data = df_cars[df_cars['transmission_type']=='Automatic'], x ='starting_price', bins = 10)
```

```python
# mean starting price of each transmission type
avg_automatic = df_cars[df_cars['transmission_type']=='Automatic']['starting_price'].mean()
avg_manual = df_cars[df_cars['transmission_type']=='Manual']['starting_price'].mean()

print(f"Mean for Automatic is: {round(avg_automatic,2)}")
print(f"Mean for Manual is: {round(avg_manual,2)}")
```

### Exercise 2 - Creating plots to find relationships between categorical & categorical data

Using plots, determine the relationship between starting_price and fuel_type / transmission_type. What are the trends?

Task:
- Create multilevel box chart of starting price by transmission_type and fuel_type
- What insights can you observe?

```python
sns.boxplot(data=df_cars, x="fuel_type", y="starting_price", hue = "transmission_type").set(title='Fuel Type vs Starting Price')
plt.show()
```

```python
# What insights can you observe?

# Manual cars generally have lower starting prices compared to automatic cars, regardless of the model.
# Electric cars do not offer a manual option, and CNG cars do not have an automatic option.
# There is a significant price difference between starting prices of automatic and manual cars.
# The black lines, which indicate the spread of the data for each category, show that petrol car prices have high variability,
#  with many outliers and a larger interquartile range.
```
